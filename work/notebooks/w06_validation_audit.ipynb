## 2. My model under an honest split (before/after)

*Re-run your Week-5 model under a grouped or time-aware split. Show both numbers.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.model_selection import train_test_split, GroupShuffleSplit
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

TARGET = "is_underperformer"

# Remove IDs and target from model features
drop_cols = ["content_id", "client_id", TARGET]

X = df.drop(columns=[c for c in drop_cols if c in df.columns])
y = df[TARGET].astype(int)

# Keep client_id separately for grouped split
groups = df["client_id"]

print("Target:", TARGET)
print("Rows:", len(df))
print("\nTarget distribution:")
print(y.value_counts())
```

```python
numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

def make_model():
    return Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ])

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

random_model = make_model()
random_model.fit(X_train_r, y_train_r)

pred_random = random_model.predict(X_test_r)

random_results = {
    "Accuracy": accuracy_score(y_test_r, pred_random),
    "Precision": precision_score(y_test_r, pred_random, zero_division=0),
    "Recall": recall_score(y_test_r, pred_random, zero_division=0),
    "F1": f1_score(y_test_r, pred_random, zero_division=0)
}

print("Random Split Results")
print(random_results)
```

## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

```python
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train_g = X.iloc[train_idx]
X_test_g = X.iloc[test_idx]

y_train_g = y.iloc[train_idx]
y_test_g = y.iloc[test_idx]

train_clients = set(groups.iloc[train_idx])
test_clients = set(groups.iloc[test_idx])

print("Train rows:", len(train_idx))
print("Test rows:", len(test_idx))
print("Train clients:", len(train_clients))
print("Test clients:", len(test_clients))
print("Client overlap:", len(train_clients.intersection(test_clients)))
```

```python
grouped_model = make_model()
grouped_model.fit(X_train_g, y_train_g)

pred_grouped = grouped_model.predict(X_test_g)

grouped_results = {
    "Accuracy": accuracy_score(y_test_g, pred_grouped),
    "Precision": precision_score(y_test_g, pred_grouped, zero_division=0),
    "Recall": recall_score(y_test_g, pred_grouped, zero_division=0),
    "F1": f1_score(y_test_g, pred_grouped, zero_division=0)
}

print("Grouped-by-client Results")
print(grouped_results)
```

```python
comparison = pd.DataFrame([
    {"Split": "Random split (before)", **random_results},
    {"Split": "Grouped by client (after)", **grouped_results}
])

display(comparison.round(3))
```

```python
# Check relationships between target and suspicious derived columns

audit_cols = [
    "health_score",
    "trend_pct",
    "needs_indexing",
    "is_quick_win",
    "needs_ctr_fix",
    "needs_engagement_fix",
    "ai_opportunity",
    "is_declining"
]

audit_cols = [c for c in audit_cols if c in df.columns]

print("Target:", TARGET)

for col in audit_cols:
    print("\n" + "=" * 50)
    print("Feature:", col)

    if df[col].dtype == "bool" or df[col].nunique(dropna=True) <= 10:
        print(pd.crosstab(df[col], df[TARGET], normalize="index").round(3))
    else:
        print(df.groupby(TARGET)[col].agg(["count", "mean", "median", "min", "max"]))
```

```python
# Look for columns that may be derived from the target
print("Unique values / cardinality:\n")

for col in df.columns:
    print(
        f"{col:30} "
        f"unique={df[col].nunique(dropna=False):6}"
    )
```

```python
# Features that may encode target-derived / post-outcome information
leakage_cols = [
    "health_score",
    "trend_pct",
    "needs_indexing",
    "is_quick_win",
    "needs_ctr_fix",
    "needs_engagement_fix",
    "ai_opportunity",
    "is_declining",
    "is_initial_refresh_candidate",
    "is_underperformer"
]

id_cols = ["content_id", "client_id"]

safe_drop = [
    c for c in leakage_cols + id_cols
    if c in df.columns
]

X_safe = df.drop(columns=safe_drop)
y_safe = df["is_underperformer"].astype(int)
groups_safe = df["client_id"]

print("Removed columns:")
print(safe_drop)

print("\nRemaining features:", X_safe.shape[1])
print("Rows:", X_safe.shape[0])
```

```python
numeric_safe = X_safe.select_dtypes(include=["number"]).columns.tolist()

categorical_safe = X_safe.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

safe_preprocessor = ColumnTransformer([
    (
        "num",
        Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ]),
        numeric_safe
    ),
    (
        "cat",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical_safe
    )
])

def make_safe_model():
    return Pipeline([
        ("preprocessor", safe_preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ])

print("Numeric:", len(numeric_safe))
print("Categorical:", len(categorical_safe))
```

```python
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(X_safe, y_safe, groups=groups_safe)
)

X_train_safe = X_safe.iloc[train_idx]
X_test_safe = X_safe.iloc[test_idx]

y_train_safe = y_safe.iloc[train_idx]
y_test_safe = y_safe.iloc[test_idx]

train_clients = set(groups_safe.iloc[train_idx])
test_clients = set(groups_safe.iloc[test_idx])

print("Train rows:", len(train_idx))
print("Test rows:", len(test_idx))
print(
    "Client overlap:",
    len(train_clients.intersection(test_clients))
)
```

```python
safe_model = make_safe_model()

safe_model.fit(X_train_safe, y_train_safe)

pred_safe = safe_model.predict(X_test_safe)

safe_results = {
    "Accuracy": accuracy_score(y_test_safe, pred_safe),
    "Precision": precision_score(
        y_test_safe, pred_safe, zero_division=0
    ),
    "Recall": recall_score(
        y_test_safe, pred_safe, zero_division=0
    ),
    "F1": f1_score(
        y_test_safe, pred_safe, zero_division=0
    )
}

print("Leakage-safe grouped validation:")
for metric, value in safe_results.items():
    print(f"{metric}: {value:.3f}")
```

```python
audit_comparison = pd.DataFrame([
    {
        "Validation": "Original random split",
        **random_results
    },
    {
        "Validation": "Original grouped split",
        **grouped_results
    },
    {
        "Validation": "Grouped + leakage audit",
        **safe_results
    }
])

display(audit_comparison.round(3))
```

### Leakage audit

Both the random split and the client-grouped split produced perfect evaluation
scores. I treat this as a reason for additional investigation rather than as
evidence that the model generalizes perfectly.

I audited the available features to check whether any feature directly or
indirectly encodes the target (`is_underperformer`). In particular, I examined
derived status, health, trend, and recommendation fields because they may have
been created using the same underlying signals as the target.

Any feature found to contain target-derived or post-outcome information will be
removed before the final validation run.

### Real failure examples

The leakage-audited grouped test split produced no misclassified examples.

I do not interpret this as proof that the model will perform perfectly on
unseen real-world data. The target may be deterministically derived from
other variables that remain in the dataset, so the observed 1.00 scores
require additional feature-provenance investigation.

Because there were no observed errors in this test split, I do not invent
failure examples. A stronger next validation step would test the model on
independently collected or future data where the target-generation process
is separated from the input features.
